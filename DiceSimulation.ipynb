# What is the Probability of Rolling an Even Number on a Six-Sided Dice?

To find the probability of an even number on a die, I will be conducting an **observational study** by collecting data on a naturally occuring situation without interfering with the results. I will be looking for the **theoretical probablity** to calculate how many times the dice should be an even number and then the **empirical probability** which shows the actual number of times that the dice was an even number. 

## Overview:

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

rng = np.random.default_rng(seed=42)

number_of_rolls = 1000
rolls = rng.integers(1, 7, size=number_of_rolls)

df = pd.DataFrame({
    "roll_number": np.arange(1, number_of_rolls + 1),
    "outcome": rolls
})
```

## Probability Calculations:

```python
df["is_even"] = df["outcome"].isin([2, 4, 6])


even_count = df["is_even"].sum()
empirical_prob = even_count / number_of_rolls


favorable_outcomes = 3  
total_outcomes = 6    
theoretical_prob = favorable_outcomes / total_outcomes
```

## Analysis:

```python
print(f"Empirical Probability of rolling an even number: {empirical_prob:.4f}")
print(f"Theoretical Probability of rolling an even number: {theoretical_prob:.4f}")
 
df["evens_so_far"] = df["is_even"].cumsum()
df["relative_freq_so_far"] = df["evens_so_far"] / df["roll_number"]
```

## Visual:

```python
plt.figure(figsize=(10, 5))
plt.plot(df["roll_number"], df["relative_freq_so_far"], label="Empirical Relative Frequency", color="#1f77b4")
plt.axhline(y=theoretical_prob, color="red", linestyle="--", label=f"Theoretical Probability ({theoretical_prob})")

plt.title("Law of Large Numbers: Rolling an Even Number on a 6-Sided Die")
plt.xlabel("Number of Rolls")
plt.ylabel("Relative Frequency of Even Numbers")
plt.ylim(0, 1)
plt.legend()
plt.grid(True, linestyle=":", alpha=0.6)
plt.show()
```

# Conclusion:

After evaluation how many times a dice would be even, the theoretical probability was 0.5 because on a six-sided die, three sides would be even numbers (2,4,6). However, the actual probability of the number of times that the die was an even number, turned out to be only 0.484, which is slightly below the estimated probability, showing that the theoretical probability isn't always accurate.
